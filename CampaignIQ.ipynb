```python
platform_analysis = df.groupby("platform").agg(
    spend = ("spend", "sum"),
    impressions = ("impressions" , "sum"),
    clicks = ("clicks", "sum"),
    conversions = ("conversions", "sum"),
    revenue = ("revenue", "sum"))
platform_analysis
```

```python
platform_analysis["CTR"] = (platform_analysis["clicks"] / platform_analysis["impressions"] * 100)
```

```python
platform_analysis["conversion_rate"] = (platform_analysis["conversions"] / platform_analysis["clicks"] * 100)
```

```python
platform_analysis["CPC"] = (platform_analysis["spend"] / platform_analysis["clicks"])
```

```python
platform_analysis["CPA"] = np.where(platform_analysis["conversions"] > 0 , platform_analysis["spend"] / platform_analysis["conversions"],0)
```

```python
platform_analysis["ROAS"] = np.where(platform_analysis["spend"] > 0 , platform_analysis["revenue"] / platform_analysis["spend"],0)
```

```python
platform_analysis.round(2)
```

```python
df["campaign_type"] = df["campaign_name"].str.extract(r"_(Brand Story|Product Launch|Promo|Educational|UGC|Influencer)_")
```

```python
monthly_analysis = df.groupby(df["date"].dt.to_period("M")).agg(
    spend = ("spend","sum"),
    conversions = ("conversions", "sum"),
    revenue = ("revenue", "sum")
)
monthly_analysis
```

```python
monthly_analysis["CPA"] = (monthly_analysis["spend"] / monthly_analysis["conversions"])
monthly_analysis["ROAS"] = (monthly_analysis["revenue"] / monthly_analysis["spend"])
monthly_analysis.round(2)
```

```python
monthly_analysis.sort_values("ROAS", ascending = False).round(2).head(10)
```

```python
monthly_analysis.sort_values("ROAS", ascending = True).round(2).head(10)
```

```python
monthly_analysis[["spend","revenue"]].plot(kind = "line", figsize = (12,6))

plt.title("Monthly Spend vs Revenue")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.xticks(rotation = 45)
plt.legend()
plt.show()
```
